```python
# ============================================================
# U-NET IMAGE SEGMENTATION
# Oxford-IIIT Pet Dataset
# TensorFlow / Keras
# ============================================================

import os
import zipfile
import tarfile
import urllib.request

import numpy as np
import matplotlib.pyplot as plt
from PIL import Image

import tensorflow as tf
from tensorflow.keras import layers, models
from tensorflow.keras.callbacks import EarlyStopping, ModelCheckpoint


# ============================================================
# 1. CONFIGURATION
# ============================================================

IMG_HEIGHT = 128
IMG_WIDTH = 128
IMG_CHANNELS = 3

BATCH_SIZE = 16

# Start with 5 epochs on CPU.
# You can change this to 10 for the final experiment.
EPOCHS = 5

SEED = 42

np.random.seed(SEED)
tf.random.set_seed(SEED)

BASE_DIR = "oxford_pet"

IMAGES_DIR = os.path.join(BASE_DIR, "images")
ANNOTATIONS_DIR = os.path.join(BASE_DIR, "annotations")
TRIMAPS_DIR = os.path.join(ANNOTATIONS_DIR, "trimaps")

OUTPUT_DIR = "outputs"

SINGLE_IMAGE_DIR = os.path.join(
    OUTPUT_DIR, "single_image_demo"
)

CURVE_DIR = os.path.join(
    OUTPUT_DIR, "training_curves"
)

PREDICTION_DIR = os.path.join(
    OUTPUT_DIR, "predictions"
)

EVALUATION_DIR = os.path.join(
    OUTPUT_DIR, "evaluation"
)

os.makedirs(SINGLE_IMAGE_DIR, exist_ok=True)
os.makedirs(CURVE_DIR, exist_ok=True)
os.makedirs(PREDICTION_DIR, exist_ok=True)
os.makedirs(EVALUATION_DIR, exist_ok=True)


# ============================================================
# 2. CHECK TENSORFLOW
# ============================================================

print("TensorFlow version:", tf.__version__)

gpus = tf.config.list_physical_devices("GPU")

if gpus:
    print("GPU detected:")
    for gpu in gpus:
        print(gpu)
else:
    print("No GPU detected. Training will use CPU.")


# ============================================================
# 3. DOWNLOAD OXFORD-IIIT PET DATASET
# ============================================================

DATASET_URL = (
    "https://www.robots.ox.ac.uk/~vgg/data/pets/"
    "data/images.tar.gz"
)

ANNOTATION_URL = (
    "https://www.robots.ox.ac.uk/~vgg/data/pets/"
    "data/annotations.tar.gz"
)

os.makedirs(BASE_DIR, exist_ok=True)

images_archive = os.path.join(
    BASE_DIR,
    "images.tar.gz"
)

annotations_archive = os.path.join(
    BASE_DIR,
    "annotations.tar.gz"
)


# ------------------------------------------------------------
# Download images
# ------------------------------------------------------------

if not os.path.exists(IMAGES_DIR):

    print("\nDownloading pet images...")

    urllib.request.urlretrieve(
        DATASET_URL,
        images_archive
    )

    print("Extracting images...")

    with tarfile.open(
        images_archive,
        "r:gz"
    ) as tar:

        tar.extractall(BASE_DIR)

    print("Images extracted.")

else:

    print("\nImages already downloaded.")


# ------------------------------------------------------------
# Download annotations
# ------------------------------------------------------------

if not os.path.exists(ANNOTATIONS_DIR):

    print("\nDownloading annotations...")

    urllib.request.urlretrieve(
        ANNOTATION_URL,
        annotations_archive
    )

    print("Extracting annotations...")

    with tarfile.open(
        annotations_archive,
        "r:gz"
    ) as tar:

        tar.extractall(BASE_DIR)

    print("Annotations extracted.")

else:

    print("\nAnnotations already downloaded.")


print("\nDataset ready.")


# ============================================================
# 4. FIND IMAGE/MASK PAIRS
# ============================================================

image_files = []
mask_files = []

valid_extensions = (
    ".jpg",
    ".jpeg",
    ".png"
)

for filename in os.listdir(IMAGES_DIR):

    if filename.lower().endswith(
        (".jpg", ".jpeg")
    ):

        image_path = os.path.join(
            IMAGES_DIR,
            filename
        )

        base_name = os.path.splitext(
            filename
        )[0]

        mask_path = os.path.join(
            TRIMAPS_DIR,
            base_name + ".png"
        )

        if os.path.exists(mask_path):

            image_files.append(
                image_path
            )

            mask_files.append(
                mask_path
            )


# ============================================================
# 5. SORT FILES
# ============================================================

combined = list(
    zip(
        image_files,
        mask_files
    )
)

combined.sort()

image_files = [
    x[0] for x in combined
]

mask_files = [
    x[1] for x in combined
]

print(
    "\nTotal image/mask pairs:",
    len(image_files)
)


# ============================================================
# 6. TRAIN / VALIDATION / TEST SPLIT
# ============================================================

total_samples = len(image_files)

indices = np.arange(total_samples)

np.random.shuffle(indices)

image_files = np.array(
    image_files
)[indices]

mask_files = np.array(
    mask_files
)[indices]


train_end = int(
    total_samples * 0.70
)

validation_end = int(
    total_samples * 0.85
)

train_images = image_files[
    :train_end
]

train_masks = mask_files[
    :train_end
]

validation_images = image_files[
    train_end:validation_end
]

validation_masks = mask_files[
    train_end:validation_end
]

test_images = image_files[
    validation_end:
]

test_masks = mask_files[
    validation_end:
]


print("\nDataset split:")
print(
    "Training:",
    len(train_images)
)

print(
    "Validation:",
    len(validation_images)
)

print(
    "Testing:",
    len(test_images)
)


# ============================================================
# 7. IMAGE PREPROCESSING FUNCTION
# ============================================================

def load_and_preprocess(
    image_path,
    mask_path
):

    # --------------------------------------------------------
    # Read image
    # --------------------------------------------------------

    image = tf.io.read_file(
        image_path
    )

    image = tf.image.decode_jpeg(
        image,
        channels=3
    )

    image = tf.image.resize(
        image,
        [
            IMG_HEIGHT,
            IMG_WIDTH
        ]
    )

    image = tf.cast(
        image,
        tf.float32
    )

    image = image / 255.0


    # --------------------------------------------------------
    # Read segmentation mask
    # --------------------------------------------------------

    mask = tf.io.read_file(
        mask_path
    )

    mask = tf.image.decode_png(
        mask,
        channels=1
    )

    # IMPORTANT:
    # Masks contain categorical labels.
    # Therefore use nearest-neighbor resizing.
    mask = tf.image.resize(
        mask,
        [
            IMG_HEIGHT,
            IMG_WIDTH
        ],
        method="nearest"
    )

    mask = tf.cast(
        mask,
        tf.int32
    )


    # --------------------------------------------------------
    # Oxford-IIIT Pet trimap labels
    #
    # 1 = foreground
    # 2 = background
    # 3 = border
    #
    # Convert to binary segmentation:
    #
    # background -> 0
    # foreground/border -> 1
    # --------------------------------------------------------

    mask = tf.where(
        mask == 2,
        0,
        1
    )

    mask = tf.cast(
        mask,
        tf.float32
    )

    return image, mask


# ============================================================
# 8. DATA AUGMENTATION
# ============================================================

def augment(
    image,
    mask
):

    random_value = tf.random.uniform(
        ()
    )

    def flip():

        return (
            tf.image.flip_left_right(
                image
            ),
            tf.image.flip_left_right(
                mask
            )
        )

    def no_flip():

        return image, mask

    image, mask = tf.cond(
        random_value > 0.5,
        flip,
        no_flip
    )

    return image, mask


# ============================================================
# 9. CREATE TF.DATA DATASETS
# ============================================================

def create_dataset(
    image_paths,
    mask_paths,
    training=False
):

    dataset = tf.data.Dataset.from_tensor_slices(
        (
            image_paths,
            mask_paths
        )
    )

    if training:

        dataset = dataset.shuffle(
            len(image_paths),
            seed=SEED
        )

    dataset = dataset.map(
        load_and_preprocess,
        num_parallel_calls=tf.data.AUTOTUNE
    )

    if training:

        dataset = dataset.map(
            augment,
            num_parallel_calls=tf.data.AUTOTUNE
        )

    dataset = dataset.batch(
        BATCH_SIZE
    )

    dataset = dataset.prefetch(
        tf.data.AUTOTUNE
    )

    return dataset


train_dataset = create_dataset(
    train_images,
    train_masks,
    training=True
)

validation_dataset = create_dataset(
    validation_images,
    validation_masks,
    training=False
)

test_dataset = create_dataset(
    test_images,
    test_masks,
    training=False
)


print("\nTensorFlow datasets created.")


# ============================================================
# 10. CHECK PREPROCESSED DATA
# ============================================================

sample_images, sample_masks = next(
    iter(train_dataset)
)

print(
    "\nImage batch shape:",
    sample_images.shape
)

print(
    "Mask batch shape:",
    sample_masks.shape
)

print(
    "Image range:",
    tf.reduce_min(
        sample_images
    ).numpy(),
    "to",
    tf.reduce_max(
        sample_images
    ).numpy()
)

print(
    "Mask values:",
    np.unique(
        sample_masks.numpy()
    )
)


# ============================================================
# 11. VISUALIZE DATASET
# ============================================================

plt.figure(
    figsize=(12, 8)
)

for i in range(4):

    plt.subplot(
        2,
        4,
        i + 1
    )

    plt.imshow(
        sample_images[i]
    )

    plt.title(
        "Input Image"
    )

    plt.axis("off")


    plt.subplot(
        2,
        4,
        i + 5
    )

    plt.imshow(
        sample_masks[i].numpy().squeeze(),
        cmap="gray"
    )

    plt.title(
        "Ground Truth Mask"
    )

    plt.axis("off")


plt.tight_layout()
plt.show()


# ============================================================
# 12. U-NET CONVOLUTION BLOCK
# ============================================================

def unet_block(
    x,
    filters
):

    x = layers.Conv2D(
        filters,
        3,
        padding="same",
        activation="relu"
    )(x)

    x = layers.Conv2D(
        filters,
        3,
        padding="same",
        activation="relu"
    )(x)

    return x


# ============================================================
# 13. BUILD U-NET
# ============================================================

def build_unet():

    inputs = layers.Input(
        shape=(
            IMG_HEIGHT,
            IMG_WIDTH,
            IMG_CHANNELS
        )
    )


    # ========================================================
    # ENCODER
    # ========================================================

    c1 = unet_block(
        inputs,
        32
    )

    p1 = layers.MaxPooling2D(
        2
    )(c1)


    c2 = unet_block(
        p1,
        64
    )

    p2 = layers.MaxPooling2D(
        2
    )(c2)


    c3 = unet_block(
        p2,
        128
    )

    p3 = layers.MaxPooling2D(
        2
    )(c3)


    # ========================================================
    # BOTTLENECK
    # ========================================================

    bottleneck = unet_block(
        p3,
        256
    )


    # ========================================================
    # DECODER
    # ========================================================

    u1 = layers.UpSampling2D(
        2
    )(bottleneck)

    u1 = layers.Concatenate()(
        [u1, c3]
    )

    c4 = unet_block(
        u1,
        128
    )


    u2 = layers.UpSampling2D(
        2
    )(c4)

    u2 = layers.Concatenate()(
        [u2, c2]
    )

    c5 = unet_block(
        u2,
        64
    )


    u3 = layers.UpSampling2D(
        2
    )(c5)

    u3 = layers.Concatenate()(
        [u3, c1]
    )

    c6 = unet_block(
        u3,
        32
    )


    # ========================================================
    # OUTPUT
    # ========================================================

    outputs = layers.Conv2D(
        1,
        1,
        activation="sigmoid"
    )(c6)


    model = models.Model(
        inputs=inputs,
        outputs=outputs,
        name="U-Net"
    )

    return model


# ============================================================
# 14. CREATE MODEL
# ============================================================

model = build_unet()

model.summary()


# ============================================================
# 15. IoU METRIC
# ============================================================

def iou_metric(
    y_true,
    y_pred
):

    y_true = tf.cast(
        y_true > 0.5,
        tf.float32
    )

    y_pred = tf.cast(
        y_pred > 0.5,
        tf.float32
    )


    intersection = tf.reduce_sum(
        y_true * y_pred
    )

    union = (
        tf.reduce_sum(y_true)
        +
        tf.reduce_sum(y_pred)
        -
        intersection
    )


    iou = (
        intersection + 1e-7
    ) / (
        union + 1e-7
    )

    return iou


# ============================================================
# 16. COMPILE MODEL
# ============================================================

model.compile(
    optimizer=tf.keras.optimizers.Adam(
        learning_rate=0.001
    ),
    loss="binary_crossentropy",
    metrics=[
        "accuracy",
        iou_metric
    ]
)

print(
    "\nModel compiled successfully."
)


# ============================================================
# 17. SINGLE IMAGE DEMONSTRATION
# ============================================================

print(
    "\nRunning single-image U-Net demonstration..."
)

demo_image = sample_images[0:1]

demo_prediction = model.predict(
    demo_image,
    verbose=0
)

demo_mask = demo_prediction[
    0
].squeeze()


plt.figure(
    figsize=(12, 4)
)


plt.subplot(
    1,
    3,
    1
)

plt.imshow(
    demo_image[0]
)

plt.title(
    "Input Image"
)

plt.axis("off")


plt.subplot(
    1,
    3,
    2
)

plt.imshow(
    demo_mask,
    cmap="gray"
)

plt.title(
    "Untrained U-Net Output"
)

plt.axis("off")


plt.subplot(
    1,
    3,
    3
)

plt.imshow(
    demo_mask > 0.5,
    cmap="gray"
)

plt.title(
    "Thresholded Mask"
)

plt.axis("off")


plt.tight_layout()


demo_path = os.path.join(
    SINGLE_IMAGE_DIR,
    "single_image_demo.png"
)

plt.savefig(
    demo_path,
    dpi=150,
    bbox_inches="tight"
)

plt.show()

print(
    "Demo saved:",
    demo_path
)


# ============================================================
# 18. MODEL CHECKPOINT
# ============================================================

MODEL_PATH = os.path.join(
    OUTPUT_DIR,
    "best_unet.keras"
)


checkpoint = ModelCheckpoint(
    MODEL_PATH,
    monitor="val_iou_metric",
    mode="max",
    save_best_only=True,
    verbose=1
)


early_stopping = EarlyStopping(
    monitor="val_iou_metric",
    mode="max",
    patience=3,
    restore_best_weights=True,
    verbose=1
)


# ============================================================
# 19. TRAIN MODEL
# ============================================================

print(
    "\nStarting U-Net training..."
)

history = model.fit(
    train_dataset,
    validation_data=validation_dataset,
    epochs=EPOCHS,
    callbacks=[
        checkpoint,
        early_stopping
    ]
)

print(
    "\nTraining completed."
)


# ============================================================
# 20. LOAD BEST MODEL
# ============================================================

model = tf.keras.models.load_model(
    MODEL_PATH,
    custom_objects={
        "iou_metric": iou_metric
    }
)

print(
    "\nBest model loaded."
)


# ============================================================
# 21. LOSS CURVE
# ============================================================

plt.figure(
    figsize=(8, 5)
)

plt.plot(
    history.history["loss"],
    label="Training Loss"
)

plt.plot(
    history.history["val_loss"],
    label="Validation Loss"
)

plt.title(
    "Training and Validation Loss"
)

plt.xlabel(
    "Epoch"
)

plt.ylabel(
    "Loss"
)

plt.legend()
plt.grid(True)

loss_path = os.path.join(
    CURVE_DIR,
    "loss_curve.png"
)

plt.savefig(
    loss_path,
    dpi=150,
    bbox_inches="tight"
)

plt.show()


# ============================================================
# 22. ACCURACY CURVE
# ============================================================

plt.figure(
    figsize=(8, 5)
)

plt.plot(
    history.history["accuracy"],
    label="Training Accuracy"
)

plt.plot(
    history.history["val_accuracy"],
    label="Validation Accuracy"
)

plt.title(
    "Training and Validation Accuracy"
)

plt.xlabel(
    "Epoch"
)

plt.ylabel(
    "Accuracy"
)

plt.legend()
plt.grid(True)

accuracy_path = os.path.join(
    CURVE_DIR,
    "accuracy_curve.png"
)

plt.savefig(
    accuracy_path,
    dpi=150,
    bbox_inches="tight"
)

plt.show()


# ============================================================
# 23. IoU CURVE
# ============================================================

plt.figure(
    figsize=(8, 5)
)

plt.plot(
    history.history["iou_metric"],
    label="Training IoU"
)

plt.plot(
    history.history["val_iou_metric"],
    label="Validation IoU"
)

plt.title(
    "Training and Validation IoU"
)

plt.xlabel(
    "Epoch"
)

plt.ylabel(
    "IoU"
)

plt.legend()
plt.grid(True)

iou_path = os.path.join(
    CURVE_DIR,
    "iou_curve.png"
)

plt.savefig(
    iou_path,
    dpi=150,
    bbox_inches="tight"
)

plt.show()


# ============================================================
# 24. TEST EVALUATION
# ============================================================

print(
    "\nEvaluating on test dataset..."
)

test_results = model.evaluate(
    test_dataset,
    verbose=1
)


print(
    "\n=============================="
)

print(
    "FINAL TEST RESULTS"
)

print(
    "=============================="
)

for name, value in zip(
    model.metrics_names,
    test_results
):

    print(
        f"{name}: {value:.4f}"
    )


test_loss = test_results[0]
test_accuracy = test_results[1]
test_iou = test_results[2]


# ============================================================
# 25. CONFUSION METRICS
# ============================================================

print(
    "\nCalculating pixel-level confusion metrics..."
)

TP = 0
TN = 0
FP = 0
FN = 0


for batch_images, batch_masks in test_dataset:

    predictions = model.predict(
        batch_images,
        verbose=0
    )


    predictions = (
        predictions > 0.5
    ).astype(
        np.uint8
    )


    true_masks = (
        batch_masks.numpy() > 0.5
    ).astype(
        np.uint8
    )


    TP += np.sum(
        (predictions == 1)
        &
        (true_masks == 1)
    )


    TN += np.sum(
        (predictions == 0)
        &
        (true_masks == 0)
    )


    FP += np.sum(
        (predictions == 1)
        &
        (true_masks == 0)
    )


    FN += np.sum(
        (predictions == 0)
        &
        (true_masks == 1)
    )


print(
    "\nTrue Positive :",
    TP
)

print(
    "True Negative :",
    TN
)

print(
    "False Positive:",
    FP
)

print(
    "False Negative:",
    FN
)


# ============================================================
# 26. PRECISION / RECALL
# ============================================================

precision = TP / (
    TP + FP + 1e-7
)

recall = TP / (
    TP + FN + 1e-7
)

pixel_accuracy = (
    TP + TN
) / (
    TP + TN + FP + FN + 1e-7
)


print(
    "\nPixel Accuracy:",
    round(pixel_accuracy, 4)
)

print(
    "Precision:",
    round(precision, 4)
)

print(
    "Recall:",
    round(recall, 4)
)


# ============================================================
# 27. SAVE CONFUSION RESULTS
# ============================================================

confusion_path = os.path.join(
    EVALUATION_DIR,
    "confusion_metrics.txt"
)


with open(
    confusion_path,
    "w"
) as file:

    file.write(
        "Pixel-Level Confusion Metrics\n"
    )

    file.write(
        "=============================\n\n"
    )

    file.write(
        f"True Positive : {TP}\n"
    )

    file.write(
        f"True Negative : {TN}\n"
    )

    file.write(
        f"False Positive: {FP}\n"
    )

    file.write(
        f"False Negative: {FN}\n\n"
    )

    file.write(
        f"Pixel Accuracy: {pixel_accuracy:.4f}\n"
    )

    file.write(
        f"Precision: {precision:.4f}\n"
    )

    file.write(
        f"Recall: {recall:.4f}\n"
    )


# ============================================================
# 28. VISUALIZE TEST PREDICTIONS
# ============================================================

print(
    "\nGenerating prediction visualizations..."
)

test_batch_images, test_batch_masks = next(
    iter(test_dataset)
)

test_predictions = model.predict(
    test_batch_images,
    verbose=0
)


NUM_SAMPLES = min(
    5,
    len(test_batch_images)
)


plt.figure(
    figsize=(
        15,
        NUM_SAMPLES * 4
    )
)


for i in range(NUM_SAMPLES):

    image = (
        test_batch_images[i]
        .numpy()
    )


    true_mask = (
        test_batch_masks[i]
        .numpy()
        .squeeze()
    )


    predicted_mask = (
        test_predictions[i]
        .squeeze()
        > 0.5
    )


    # --------------------------------------------------------
    # Input
    # --------------------------------------------------------

    plt.subplot(
        NUM_SAMPLES,
        3,
        i * 3 + 1
    )

    plt.imshow(
        image
    )

    plt.title(
        "Input Image"
    )

    plt.axis("off")


    # --------------------------------------------------------
    # Ground Truth
    # --------------------------------------------------------

    plt.subplot(
        NUM_SAMPLES,
        3,
        i * 3 + 2
    )

    plt.imshow(
        true_mask,
        cmap="gray"
    )

    plt.title(
        "Ground Truth Mask"
    )

    plt.axis("off")


    # --------------------------------------------------------
    # Prediction
    # --------------------------------------------------------

    plt.subplot(
        NUM_SAMPLES,
        3,
        i * 3 + 3
    )

    plt.imshow(
        predicted_mask,
        cmap="gray"
    )

    plt.title(
        "Predicted Mask"
    )

    plt.axis("off")


plt.tight_layout()


prediction_path = os.path.join(
    PREDICTION_DIR,
    "test_predictions.png"
)


plt.savefig(
    prediction_path,
    dpi=150,
    bbox_inches="tight"
)

plt.show()


print(
    "Prediction image saved:",
    prediction_path
)


# ============================================================
# 29. SAVE INDIVIDUAL PREDICTIONS
# ============================================================

for i in range(NUM_SAMPLES):

    image = (
        test_batch_images[i]
        .numpy()
    )

    true_mask = (
        test_batch_masks[i]
        .numpy()
        .squeeze()
    )

    predicted_mask = (
        test_predictions[i]
        .squeeze()
        > 0.5
    )


    plt.figure(
        figsize=(12, 4)
    )


    plt.subplot(
        1,
        3,
        1
    )

    plt.imshow(
        image
    )

    plt.title(
        "Original Image"
    )

    plt.axis("off")


    plt.subplot(
        1,
        3,
        2
    )

    plt.imshow(
        true_mask,
        cmap="gray"
    )

    plt.title(
        "Ground Truth"
    )

    plt.axis("off")


    plt.subplot(
        1,
        3,
        3
    )

    plt.imshow(
        predicted_mask,
        cmap="gray"
    )

    plt.title(
        "Prediction"
    )

    plt.axis("off")


    plt.tight_layout()


    individual_path = os.path.join(
        PREDICTION_DIR,
        f"sample_{i + 1}.png"
    )


    plt.savefig(
        individual_path,
        dpi=150,
        bbox_inches="tight"
    )

    plt.close()


# ============================================================
# 30. SAVE MODEL
# ============================================================

FINAL_MODEL_PATH = os.path.join(
    OUTPUT_DIR,
    "final_unet_model.keras"
)


model.save(
    FINAL_MODEL_PATH
)


print(
    "\nModel saved:",
    FINAL_MODEL_PATH
)


# ============================================================
# 31. SAVE RESULTS SUMMARY
# ============================================================

results_path = os.path.join(
    OUTPUT_DIR,
    "results_summary.txt"
)


with open(
    results_path,
    "w"
) as file:

    file.write(
        "U-NET OXFORD-IIIT PET SEGMENTATION\n"
    )

    file.write(
        "===================================\n\n"
    )

    file.write(
        f"Image Size: "
        f"{IMG_HEIGHT} x {IMG_WIDTH}\n"
    )

    file.write(
        f"Batch Size: {BATCH_SIZE}\n"
    )

    file.write(
        f"Epochs: {EPOCHS}\n\n"
    )

    file.write(
        f"Test Loss: "
        f"{test_loss:.4f}\n"
    )

    file.write(
        f"Test Accuracy: "
        f"{test_accuracy:.4f}\n"
    )

    file.write(
        f"Test IoU: "
        f"{test_iou:.4f}\n\n"
    )

    file.write(
        f"True Positive: {TP}\n"
    )

    file.write(
        f"True Negative: {TN}\n"
    )

    file.write(
        f"False Positive: {FP}\n"
    )

    file.write(
        f"False Negative: {FN}\n\n"
    )

    file.write(
        f"Pixel Accuracy: "
        f"{pixel_accuracy:.4f}\n"
    )

    file.write(
        f"Precision: "
        f"{precision:.4f}\n"
    )

    file.write(
        f"Recall: "
        f"{recall:.4f}\n"
    )


# ============================================================
# 32. FINAL OUTPUT
# ============================================================

print("\n")
print("=" * 60)
print("U-NET LAB COMPLETED")
print("=" * 60)

print("\nFinal Test Results:")
print(
    f"Loss     : {test_loss:.4f}"
)

print(
    f"Accuracy : {test_accuracy:.4f}"
)

print(
    f"IoU      : {test_iou:.4f}"
)

print("\nConfusion Metrics:")
print(
    f"TP : {TP}"
)

print(
    f"TN : {TN}"
)

print(
    f"FP : {FP}"
)

print(
    f"FN : {FN}"
)

print("\nGenerated files:")

print(
    "\nSingle Image Demo:"
)

print(
    demo_path
)

print(
    "\nTraining Curves:"
)

print(
    loss_path
)

print(
    accuracy_path
)

print(
    iou_path
)

print(
    "\nPredictions:"
)

print(
    prediction_path
)

print(
    "\nEvaluation:"
)

print(
    confusion_path
)

print(
    "\nResults:"
)

print(
    results_path
)

print(
    "\nModel:"
)

print(
    FINAL_MODEL_PATH
)

print(
    "\nAll tasks completed."
)
```
